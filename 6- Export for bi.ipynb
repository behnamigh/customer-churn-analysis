```python
import pandas as pd
```

```python
df = pd.read_csv('cleaned_telco.csv')
```

```python
# Prepare a clean, BI-ready dataset with derived fields for Power BI
df['Tenure Bucket'] = pd.cut(df['Tenure in Months'], bins=[0,12,24,48,72], 
                               labels=['0-12','13-24','25-48','49-72'])
df['At Risk Flag'] = ((df['CLTV'] > 5000) & (df['Churn Score'] > 70) & (df['Churned']==0)).map({True:'High Risk', False:'Normal'})

bi_cols = ['Customer ID','Gender','Age','Senior Citizen','Married','Dependents',
           'City','State','Tenure in Months','Tenure Bucket','Contract','Internet Type',
           'Payment Method','Monthly Charge','Total Charges','Total Revenue',
           'Satisfaction Score','Churn Score','CLTV','Churned','Churn Label',
           'Churn Category','Churn Reason','At Risk Flag']

df[bi_cols].to_csv('powerbi_dataset.csv', index=False)
print("Power BI-ready dataset exported:", df[bi_cols].shape)
print(df['At Risk Flag'].value_counts())
```
