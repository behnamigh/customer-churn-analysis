```python
import pandas as pd
```

```python
df = pd.read_csv('cleaned_telco.csv')
```

```python
print("=== Churn rate by Contract and Internet Type ===")
q1 = (
    df.groupby(['Contract', 'Internet Type'])['Churned']
    .agg(total_customers='count', churned_customers='sum')
)
q1['churn_rate_pct'] = (100 * q1['churned_customers'] / q1['total_customers']).round(2)
q1 = q1.sort_values('churn_rate_pct', ascending=False).head(10)
print(q1)
```

```python
print("\n=== High-value customers at risk ===")
q2 = df[
    (df['Churned'] == 0) &
    (df['CLTV'] > 5000) &
    (df['Churn Score'] > 70)
][['Customer ID', 'CLTV', 'Churn Score', 'Monthly Charge', 'Contract']]
q2 = q2.sort_values('CLTV', ascending=False).head(10)
print(q2)
```

```python
print("\n=== Revenue lost by churn category ===")
q3 = (
    df[df['Churned'] == 1]
    .groupby('Churn Category')['Total Revenue']
    .agg(customers_lost='count', revenue_lost='sum')
)
q3['revenue_lost'] = q3['revenue_lost'].round(2)
q3 = q3.sort_values('revenue_lost', ascending=False)
print(q3)
```

```python
with open('business_questions_results.txt', 'w', encoding='utf-8') as f:
    f.write("=== Churn rate by Contract and Internet Type ===\n")
    f.write(q1.to_string())
    f.write("\n\n=== High-value customers at risk ===\n")
    f.write(q2.to_string(index=False))
    f.write("\n\n=== Revenue lost by churn category ===\n")
    f.write(q3.to_string())
```
