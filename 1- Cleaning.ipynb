```python
import pandas as pd
import numpy as np
```

```python
df =pd.read_csv('telco.csv')
```

```python
df = df[df['Customer Status'] != 'Joined']
```

```python
df['Churn Category'] = df['Churn Category'].fillna('None')
df['Churn Reason'] = df['Churn Reason'].fillna('None')
```

```python
df['Churned'] = (df['Churn Label'] == 'Yes').astype(int)
```

```python
df.to_csv('cleaned_telco.csv', index=False)
```
