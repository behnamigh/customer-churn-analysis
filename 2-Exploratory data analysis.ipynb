```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('cleaned_telco.csv')
```

```python
print("=== Churn rate by Contract ===")
print(df.groupby('Contract')['Churned'].agg(['mean','count']).sort_values('mean', ascending=False))
```

```python
print("=== Churn rate by Internet Type ===")
print(df.groupby('Internet Type')['Churned'].agg(['mean','count']).sort_values('mean', ascending=False))
```

```python
print("=== Churn rate by Payment Method ===")
print(df.groupby('Payment Method')['Churned'].agg(['mean','count']).sort_values('mean', ascending=False))
```

```python
print("=== Churn by Tenure buckets ===")
df['Tenure Bucket'] = pd.cut(df['Tenure in Months'], bins=[0,12,24,48,72], labels=['0-12','13-24','25-48','49-72'])
print(df.groupby('Tenure Bucket', observed=True)['Churned'].agg(['mean','count']))
```

```python
print("=== Top Churn Reasons (among churned) ===")
print(df[df['Churned']==1]['Churn Reason'].value_counts().head(10))
```

```python
print("=== Top Churn Categories ===")
print(df[df['Churned']==1]['Churn Category'].value_counts())
```

```python
print("=== Avg Monthly Charge: Churned vs Stayed ===")
print(df.groupby('Churned')['Monthly Charge'].mean())
```

```python
print("=== Satisfaction Score vs Churn ===")
print(df.groupby('Churned')['Satisfaction Score'].mean())
```
